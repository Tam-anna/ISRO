```python
# Install ultralytics
!pip install ultralytics

# Mount Google Drive
from google.colab import drive
drive.mount('/content/drive')

from ultralytics import YOLO

# Load YOLOv11n model
model = YOLO('yolo11n.pt')

# Train model for 200 epochs and save results in Drive
model.train(
    data='/content/Illegal/Intrusion+border/data.yaml',
    epochs=200,
    imgsz=640,
    project='/content/drive/MyDrive/Illegal_Detection_Project',  # save in Drive
    name='yolo11n_train_200'
)
```
